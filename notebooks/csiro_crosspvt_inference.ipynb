# CrossPVT Fold Ensemble Submission (Scripted)

This notebook simply configures paths and executes scripts/run_crosspvt_inference.py, so all preprocessing and inference logic stays inside the uploaded src/ & scripts/ directories. Attach the dataset that contains these folders (e.g. csiro-vitonly) plus the competition data (csiro-biomass), adjust the constants below if you renamed anything, then run the notebook to produce submission.csv.


```python
from pathlib import Path
import subprocess

CHECKPOINT_DATASET = Path("/kaggle/input/csiro-vitonly")
RUN_NAME = "crosspvt_imgauto_bs8_lr0.0001_20260113_130321"
COMPETITION_DATASET = Path("/kaggle/input/csiro-biomass")
FOLDS = [0, 1, 2, 3, 4]
CKPT_FILENAME = "best_wr2.pt"
OUTPUT_PATH = Path("submission.csv")

script_path = CHECKPOINT_DATASET / "scripts" / "run_crosspvt_inference.py"
cmd = [
    "python",
    str(script_path),
    "--checkpoint-root",
    str(CHECKPOINT_DATASET),
    "--run-name",
    RUN_NAME,
    "--competition-root",
    str(COMPETITION_DATASET),
    "--folds",
    *[str(f) for f in FOLDS],
    "--ckpt-filename",
    CKPT_FILENAME,
    "--output",
    str(OUTPUT_PATH),
]
print("Running:", " ".join(cmd))
subprocess.run(cmd, check=True)
print(f"submission saved to {OUTPUT_PATH.resolve()}")
```
